### 1. Write a function that takes a list value as an argument and returns a string with all the items separated by a comma and a space, with 'and' inserted before the last item. (e.g. ['apples', 'bananas', 'tofu', 'cats'] -> 'apples, bananas, tofu, and cats')

```python
def list_to_string(items):
    if len(items) == 0:
        return ""
    if len(items) == 1:
        return items[0]
    return ", ".join(items[:-1]) + ", and " + items[-1]

spam = ['apples', 'bananas', 'tofu', 'cats']
print(list_to_string(spam))
```

### 2. ARS Gems Store: calculate the bill amount for a customer based on the gems and quantity purchased. A total above Rs.30000 gets a 5% discount. If any required gem is not available in the store, the total bill amount is -1. Perform case-sensitive comparison.

```python
gems = {
    "Ruby": 10000,
    "Diamond": 50000,
    "Emerald": 20000,
    "Sapphire": 15000
}

def calculate_bill(gems, customer_list):
    
    total = 0
    
    for i in range(0, len(customer_list), 2):
        gem = customer_list[i]
        quantity = customer_list[i + 1]
        
        if gem not in gems:
            return -1
        
        total = total + gems[gem] * quantity
    
    if total > 30000:
        total = total - (total * 5 / 100)
        
    return total


customer_list = ["Ruby", 2, "Emerald", 2]

print(calculate_bill(gems, customer_list))
```

### 4. WAP to find the maximum number from num1 to num2 (num2 inclusive) whose digit sum is a multiple of 3, which has only two digits, and which is a multiple of 5. num1 must be less than num2. Display -1 for invalid data or if no such number exists.

```python
def find_max(num1, num2):
    if num1 >= num2:
        return -1

    result = []

    for num in range(num1, num2 + 1):

        # Number should have exactly two digits
        if 10 <= num <= 99:

            # Sum of digits
            digit_sum = (num // 10) + (num % 10)

            # Check all conditions
            if digit_sum % 3 == 0 and num % 5 == 0:
                result.append(num)

    if len(result) == 0:
        return -1

    return max(result)


print(find_max(10, 50))
print(find_max(20, 30))
print(find_max(50, 50))
```

### 5. WAP to generate ticket numbers in the format airline:src:dest:number, where airline is AI, src and dest are the first three characters of the source and destination cities, and number starts from 101. Return the ticket numbers of the last five passengers (or all of them if there are fewer than 5).

```python
def generate_ticket(source, destination, passenger_count):
    tickets = []

    for i in range(passenger_count):
        ticket_number = (
            "AI:"
            + source[:3].upper()
            + ":"
            + destination[:3].upper()
            + ":"
            + str(101 + i)
        )
        tickets.append(ticket_number)

    if passenger_count > 5:
        return tickets[-5:]

    return tickets


# Test cases
print(generate_ticket("Delhi", "Mumbai", 3))
print(generate_ticket("Chandigarh", "Bangalore", 7))
```
