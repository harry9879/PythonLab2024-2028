2.	Write a python program to calculate the frequency of vowels and consonants in file.

```python
vowels = set("aeiouAEIOU")
vowel_count = 0
consonant_count = 0

file = open("test.txt", "r")
text = file.read()
file.close()

for character in text:
    if character.isalpha():
        if character in vowels:
            vowel_count += 1
        else:
            consonant_count += 1

print("Frequency of vowels:", vowel_count)
print("Frequency of consonants:", consonant_count)
```

3.	Write a python program to swap the first and second half of the input file content.

```python
file = open("test.txt", "r")
content = file.read()

midpoint = len(content) // 2
swapped_content = content[midpoint:] + content[:midpoint]

print("Content after swapping the two halves:")
print(swapped_content)
file.close()
```

4.	Write a program to copy lines which start with a lowercase letter only from the input file Demo.txt and ignore the lines which start from with an uppercase letter. The output file Demo2.txt should contain only those lines from the file Demo.txt which start with a lowercase letter.

```python
input_file = open("Demo.txt", "r")
lines = input_file.readlines()

lowercase_lines = [line for line in lines if line and line[0].islower()]

output_file = open("Demo2.txt", "w")
output_file.writelines(lowercase_lines)

print("Lines starting with lowercase letters copied to Demo2.txt:", len(lowercase_lines))
input_file.close()
output_file.close()
```
