### 1. WAP to categorize all the elements of a given tuple according to their data type. Store the results of each category in a new sub-tuple and display each sub-tuple in sorted order.

```python
t = (10, 3.5, "Apple", 25, "Banana", 7.2, True, 5, "Mango")

integers = ()
floats = ()
strings = ()
booleans = ()

for item in t: 
    if type(item) == int:
        integers += (item,)
    elif type(item) == float:
        floats += (item,)
    elif type(item) == str:
        strings += (item,)
    elif type(item) == bool:
        booleans += (item,)

# Sort each sub-tuple
integers = tuple(sorted(integers))
floats = tuple(sorted(floats))
strings = tuple(sorted(strings))
booleans = tuple(sorted(booleans))

print("Original tuple:", t)
print("Integers tuple:", integers)
print("Float tuple:", floats)
print("String tuple:", strings)
print("Boolean tuple:", booleans)
```

### 3. Care hospital: given a list of patient ids and the medical speciality each visited, and the dictionary {"P":"Pediatrics","O":"Orthopedics","E":"ENT"}, write a function to find the speciality visited by the maximum number of patients and return its name. Perform case-sensitive comparison.

```python
def find_max_speciality(patient_list):
    speciality = {
        "P": "Pediatrics",
        "O": "Orthopedics",
        "E": "ENT"
    }

    count = {
        "P": 0,
        "O": 0,
        "E": 0
    }
    
    for i in range(1,len(patient_list),2):
        code = patient_list[i]
        count[code] += 1

    max_count = max(count, key=count.get) # type: ignore
    return speciality[max_count]


patients = [101, "P", 102, "O", 302, "P", 305, "P"]
print(find_max_speciality(patients))
```

### 4. Write a function find_correct() which accepts a dictionary of correct spellings (keys) and contestant spellings (values) and returns a list with the number of CORRECT (exact match), ALMOST CORRECT (no more than 2 letters wrong) and WRONG (more than 2 letters wrong or length mismatch) answers.

```python
def find_correct(words):
    correct = 0
    almost_correct = 0
    wrong = 0
    
    for actual, given in words.items():
        
        if len(actual) != len(given):
            wrong+=1
            continue
        
        mistake = 0 
        for i in range(len(actual)):
            if actual[i] != given[i]:
                mistake += 1
        
        # Categorize the answer
        
        if mistake == 0:
            correct +=1
        elif mistake <= 2:
            almost_correct += 1 
        else:
            wrong+=1
        
    
    return [correct, almost_correct, wrong]


words = {
    "THEIR": "THEIR",
    "BUSINESS": "BISINESS",
    "WINDOWS": "WINDMILL",
    "WERE": "WEAR",
    "SAMPLE": "SAMPLE"
}

print(find_correct(words))
```

### 5. (a) Create a character translator tr(srcstr, dststr, string) that works like the Unix tr command, translating each character in srcstr to the matching character in dststr. (e.g. tr('abc', 'mno', 'abcdef') -> 'mnodef')

```python
def tr(srcstr, dststr, string):
    result = ""
    
    for ch in string:
        if ch in srcstr:
            index = srcstr.index(ch)
            result += dststr[index]
        else:
            result += ch
    return result

print(tr("abc", "mno", "abcdef"))
```

### 5. (b) Update tr() to handle deletions: any extra characters in srcstr beyond those mapped to dststr are removed from the string. (e.g. tr('abcdef', 'mno', 'abcdefghi') -> 'mnoghi')

```python
def tr(srcstr, dststr, string):
    result = ""
    
    for ch in string:
        if ch in srcstr:
            index = srcstr.index(ch)
        
            if index < len(dststr):
                result += dststr[index]
            # Otherwise, don't add the character
        else:
            result += ch
            
    return result 

print(tr("abcdef", "mno", "abcdefghi"))
        
            
```
