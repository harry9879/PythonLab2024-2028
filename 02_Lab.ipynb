### 1. WAP to find the number of days in a month.

```python
def days_in_month(month, year):
    if month == 2:
        if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
            return 29
        return 28
    elif month in (4, 6, 9, 11):
        return 30
    elif 1 <= month <= 12:
        return 31
    else:
        return -1

month = 2
year = 2024
days = days_in_month(month, year)
if days == -1:
    print("Invalid month")
else:
    print(f"Month {month} of {year} has {days} days")
```

### 2. WAP to display the pattern of numbers

```python
n = 4

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()

for i in range(n - 1, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

### 3. WAP using while loop, which prints the sum of every fifth number from 0 to 500 (including both 0 and 500)

```python
num = 0
total = 0

while num <= 500:
    total += num
    num += 5

print("Sum:", total)
```

### 4. Display all the common characters between two strings. Return -1 if there are no matching characters.

```python
def common_characters(str1, str2):
    common = ""
    for ch in str1:
        if ch in str2 and ch not in common:
            common += ch
    if common == "":
        return -1
    return common

print(common_characters("python", "typhoon"))
print(common_characters("abc", "xyz"))
```

### 5. Merge first element of list1 with last element of list2, second with second last, and so on. If an element is None, keep the other one as it is.

```python
def merge_lists(list1, list2):
    merged = []
    for i in range(len(list1)):
        first = list1[i]
        second = list2[len(list2) - 1 - i]
        if first is None:
            merged.append(second)
        elif second is None:
            merged.append(first)
        else:
            merged.append(first + second)
    return " ".join(merged)

list1 = ['A', 'app', 'a', 'd', 'ke', 'th', 'doc', 'awa']
list2 = ['y', 'tor', 'e', 'eps', 'ay', None, 'le', 'n']
print(merge_lists(list1, list2))
```
