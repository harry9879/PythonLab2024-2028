### 1. Compress a string of uppercase characters (A-Z) using Run Length Encoding, replacing each run of a repeated character with the length of that run. (e.g. AAAABBBBCCCCCCCC -> 4A4B8C)

```python
def run_length_encode(s):

    if not s:
        return ""

    count = 1
    result = ""

    for i in range(1, len(s)):
        if (s[i] == s[i-1]):
            count +=1
        else:
            result += str(count) + s[i-1]
            count = 1

    result += str(count) + s[-1]

    return result
print(run_length_encode("AAAABBBBCCCCCCCC"))
print(run_length_encode("AABCCA"))
print(run_length_encode("AAAAA"))
print(run_length_encode("ABCD"))
```

### 2. Write a function find_pairs_of_numbers() which accepts a list of positive integers with no repetitions and returns the count of pairs of numbers in the list that add up to n. Return 0 if no such pairs are found.

```python
def find_pairs_of_numbers(lst, n):
    count = 0

    for i in range(len(lst)):
        for j in range(i+1, len(lst)):
            if lst[i] + lst[j] == n:
                count += 1

    return count
# Test cases
print(find_pairs_of_numbers([1, 2, 7, 4, 5, 6, 0, 3], 6))
print(find_pairs_of_numbers([3, 4, 1, 8, 5, 9, 0, 6], 9))
```

### 3. Using sample_data = range(1,11), create functions odd() and even() that return lists of the odd and even numbers. Create sum_of_numbers() which accepts the sample data and/or a function: if no function is passed, return the sum of all numbers; otherwise return the sum of the numbers returned by that function.

```python
sample_data = range(1,11)

def odd(data):
    return [x for x in data if x % 2 != 0]

def even(data):
    return [x for x in data if x % 2 == 0]


def sum_of_numbers(data, func = None):
    if func is None:
        return sum(data)
    else:
        return sum(func(data))

print(sum_of_numbers(sample_data))
print(sum_of_numbers(sample_data, odd))
print(sum_of_numbers(sample_data, even))
```

### 4. Marks of 10 students (out of 25) are stored in a tuple. Write functions: find_more_than_average() to return the percentage of students scoring above the class average, generate_frequency() to return a list of how many students scored each mark from 0 to 25, and sort_marks() to return the marks sorted in increasing order.

```python
def find_more_than_average(marks):
    average = sum(marks) / len(marks)
    count = 0

    for mark in marks:
        if mark > average:
            count += 1

    percentage = (count / len(marks)) * 100
    return percentage


def generate_frequency(marks):
    frequency = [0] * 26

    for mark in marks:
        frequency[mark] += 1

    return frequency


def sort_marks(marks):
    return sorted(marks)


# Sample data
marks = (12, 18, 15, 20, 12, 10, 18, 25, 15, 8)

print("Percentage scoring more than average:", find_more_than_average(marks))

print("Frequency:")
print(generate_frequency(marks))

print("Sorted marks:", sort_marks(marks))
```

### 5. Write a function check_anagram() which accepts two strings and returns True if one is an anagram of the other, otherwise False. The strings must be the same length and no character may repeat at the same position. Perform case-insensitive comparison.

```python
def check_anagram(str1, str2):
    str1 = str1.lower()
    str2 = str2.lower()

    if len(str1) != len(str2):
        return False

    for i in range(len(str1)):
        if str1[i] == str2[i]:
            return False

    if sorted(str1) == sorted(str2):
        return True
    return False

print(check_anagram("eat", "tea"))
print(check_anagram("backward", "drawback"))
print(check_anagram("Reductions", "discounter"))
print(check_anagram("About", "table"))
```
